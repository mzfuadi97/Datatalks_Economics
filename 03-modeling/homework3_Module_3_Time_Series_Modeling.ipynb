## 0.4) [Code snippet 3] Temporal split of ~25 years of data (by date)

```python
def temporal_split(df, min_date, max_date, train_prop=0.7, val_prop=0.15, test_prop=0.15):
    """
    Splits a DataFrame into three buckets based on the temporal order of the 'Date' column.

    Args:
        df (DataFrame): The DataFrame to split.
        min_date (str or Timestamp): Minimum date in the DataFrame.
        max_date (str or Timestamp): Maximum date in the DataFrame.
        train_prop (float): Proportion of data for training set (default: 0.6).
        val_prop (float): Proportion of data for validation set (default: 0.2).
        test_prop (float): Proportion of data for test set (default: 0.2).

    Returns:
        DataFrame: The input DataFrame with a new column 'split' indicating the split for each row.
    """
    # Define the date intervals
    train_end = min_date + pd.Timedelta(days=(max_date - min_date).days * train_prop)
    val_end = train_end + pd.Timedelta(days=(max_date - min_date).days * val_prop)

    # Assign split labels based on date ranges
    split_labels = []
    for date in df['Date']:
        if date <= train_end:
            split_labels.append('train')
        elif date <= val_end:
            split_labels.append('validation')
        else:
            split_labels.append('test')

    # Add 'split' column to the DataFrame
    df['split'] = split_labels

    return df
```

```python
min_date_df = df_with_dummies.Date.min()
max_date_df = df_with_dummies.Date.max()

df_with_dummies = temporal_split(df_with_dummies,
                                 min_date = min_date_df,
                                 max_date = max_date_df)
```

```python
# Decision Tree doesn't like too large and inf. values
import numpy as np

def remove_infinite_values(X):
    """
    Remove infinite values from the input array.

    Parameters:
    - X: Input array (NumPy array or array-like)

    Returns:
    - Array with infinite values removed
    """
    return X[np.isfinite(X).all(axis=1)]

# Example usage:
# Assuming X is your input data
# filtered_X = remove_infinite_values(X)
```

```python
# you may want to remove 1-2% outliers based on percentile ==> not used here in Decision Trees
def remove_outliers_percentile(X, lower_percentile=1, upper_percentile=99):
    """
    Remove outliers from the input array based on percentiles.

    Parameters:
    - X: Input array (NumPy array or array-like)
    - lower_percentile: Lower percentile threshold (float, default=1)
    - upper_percentile: Upper percentile threshold (float, default=99)

    Returns:
    - Array with outliers removed
    """
    lower_bound = np.percentile(X, lower_percentile, axis=0)
    upper_bound = np.percentile(X, upper_percentile, axis=0)
    mask = np.logical_and(np.all(X >= lower_bound, axis=1), np.all(X <= upper_bound, axis=1))
    return X[mask]

# Example usage:
# Assuming X is your input data
# filtered_X = remove_outliers_percentile(X, lower_percentile=1, upper_percentile=99)
```

### 1.4.3 Inference for a Decision Tree

```python
def predict_decision_tree(clf:DecisionTreeClassifier, df_X:pd.DataFrame, y_true: pd.Series):
  # Predict the target variable on the test data
  y_pred = clf.predict(df_X)

  max_depth = clf.tree_.max_depth
  # Print the maximum depth
  print("Maximum depth of the decision tree:", max_depth)

  # Calculate the accuracy/precision of the model
  accuracy = accuracy_score(y_test, y_pred)
  precision = precision_score(y_test, y_pred)
  print(f'Accuracy ={accuracy}, precision = {precision}')

  # resulting df
  result_df = pd.concat([df_X, y_true, pd.Series(y_pred, index=df_X.index, name='pred_')], axis=1)

  return result_df
```

### Advanced: generic function for a uniquely correct prediction

```python
def is_uniquely_correct(row, pred_col, other_pred_cols, target='is_positive_growth_30d_future'):
    """1 if `pred_col` is correct while ALL `other_pred_cols` are incorrect"""
    if row[pred_col] != row[target]:
        return 0
    return int(all(row[c] != row[target] for c in other_pred_cols))

hand_preds = ['pred0_manual_cci', 'pred1_manual_prev_g1', 'pred2_manual_prev_g1_and_snp',
              'pred3_manual_dgs10_5', 'pred4_manual_dgs10_fedfunds']
test_rows = new_df[new_df.split=='test']
test_rows.apply(lambda r: is_uniquely_correct(r, 'pred5_clf_10', hand_preds), axis=1).sum()
```

## [HOMEWORK] Question 4: Hyperparameter tuning for a Decision Tree (`max_depth` 1..20)

```python
%%time
from sklearn.tree import export_text

X_train_fit = X_train_imputed.drop(['Date','Ticker'], axis=1)
X_test_fit = X_test_imputed.drop(['Date','Ticker'], axis=1)

# optional: VALIDATION tracking (fit on TRAIN only, evaluate on VALIDATION) to observe overfitting
train_only_mask = (new_df.split == 'train').values
val_mask = (new_df.split == 'validation').values

results = []
for max_depth in range(1, 21):
    # main: fit on TRAIN+VALIDATION, precision on TEST
    clf, _ = fit_decision_tree(X=X_train_fit, y=y_train, max_depth=max_depth)
    y_pred_test = clf.predict(X_test_fit)
    prec_test = precision_score(y_test, y_pred_test, zero_division=0)
    acc_test = accuracy_score(y_test, y_pred_test)

    # tracking: fit on TRAIN only, precision on VALIDATION
    clf_tr, _ = fit_decision_tree(X=X_all[train_only_mask], y=y_all[train_only_mask], max_depth=max_depth)
    prec_val = precision_score(y_all[val_mask], clf_tr.predict(X_all[val_mask]), zero_division=0)

    results.append({'max_depth': max_depth, 'precision_test': prec_test, 'accuracy_test': acc_test,
                    'precision_validation': prec_val, 'n_positive_test': int(y_pred_test.sum())})
    print(f'max_depth={max_depth:2d}: precision TEST={prec_test:.4f}, accuracy TEST={acc_test:.4f}, precision VALID={prec_val:.4f}')

    if max_depth in (1, 2, 3):
        print(export_text(clf, feature_names=list(X_train_fit), max_depth=3))

results_df = pd.DataFrame(results).set_index('max_depth')
```

## [HOMEWORK] Question 5 (exploratory): What data is missing?

The correlation analysis and the trees are dominated by **macro/rates variables** (`DGS10`, `DGS5`, `FEDFUNDS`, `DGS1`), **long-horizon growth** (`growth_365d`, `growth_snp500_*`) and **seasonality** (`Month`, `month_wom`). The model mostly learns the *market regime*, not stock-specific information. Ideas to add:

1. **Yield-curve shape & real rates** – spreads `DGS10-DGS2`, `DGS10-DGS3MO` (FRED: `T10Y2Y`, `T10Y3M`) and 10Y real yield (`DFII10`). Level thresholds like `DGS10 <= 4` are regime-dependent; spreads and changes (Δ over 30/90 days) generalise better across periods.
2. **Risk sentiment / volatility** – VIX (`^VIX` on Yahoo Finance), India VIX (`^INDIAVIX`), VSTOXX; credit spreads (FRED `BAMLH0A0HYM2`). Future 30d returns strongly depend on risk regime.
3. **Local macro for non-US stocks** – the dataset has India and EU stocks but macro is US-only: add RBI repo rate, ECB deposit rate, Bund 10Y yield, INR/USD and EUR/USD FX (Yahoo `INR=X`, `EURUSD=X`), local indices (`^NSEI`, `^STOXX50E`) growth.
4. **Company fundamentals & events** – earnings dates and surprises, P/E, EPS revisions (yfinance `Ticker.earnings_dates`, `Ticker.info`), days-to-next-earnings, dividends/splits.
5. **Relative / cross-sectional features** – stock growth relative to its index/sector (e.g. `growth_30d - growth_snp500_30d`), rank of momentum among all 33 tickers on the same day, sector ETF growth (`XLK`, `XLF`, ...).
6. **Commodities & dollar** – oil (`CL=F`), gold (`GC=F`), DXY (`DX-Y.NYB`), Bitcoin – regime indicators especially for energy (India/EU) stocks.
7. **Calendar events** – FOMC meeting days, options expiry week (3rd Friday), month-end/quarter-end flags, pre-holiday days – a more economically-grounded version of the `month_wom` dummies.

Methodologically: use **differences/percent changes** of macro variables instead of levels (levels create non-stationary thresholds, e.g. `DGS10 <= 4.825` yields no positives on TEST), and prefer walk-forward validation.
